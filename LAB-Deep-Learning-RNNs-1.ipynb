### 3.2 Create Time Sequences
We will transform the data into sequences where each sequence represents a window of 60 days of stock prices, and the target is the price on the 61st day.

**Tasks**:

- Create a function to generate sequences of stock prices.

```python
import numpy as np

def create_sequences(data, time_steps=60):
    sequences = []
    targets = []
    for i in range(len(data) - time_steps):
        sequences.append(data[i:i + time_steps])
        targets.append(data[i + time_steps])
    return np.array(sequences), np.array(targets)

close_prices = data['Close'].values
X, y = create_sequences(close_prices, 60)
```
